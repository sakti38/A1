## **Vector**

vector adalah objek yang memiliki besar (magnitude) dan arah (direction)

makna sebenarnya bukan sekadar angka, melainkan perpindahan (displacement) dari satu titik ke titik lain

Berbeda dengan titik yang hanya suatu titik, vector itu bergerak misal v = (2, 3) artinya “Bergerak 2 ke kanan dan 3 ke atas”
- Norm

    Norm adalah ukuran panjang atau besar (magnitude) dari suatu vektor atau panjang panah

    $$ ||v|| = \sqrt{v_1^2 + v_2^2 + ... + v_n^2} $$
- Normalisasi

    Mengubah vektor agar panjangnya menjadi 1
    $$V_{\text{Normalisasi}} = \frac{v}{||v||} $$

#### **Operasi**

- Dot Product

    “Seberapa searah dua vektor”

    Seberapa besar vektor a “searah” dengan vektor b

    Jika searah maka nilainya besar postif, namun jika tegak lurus maka akan 0, sedangkan jika berlawanan arah nilainya besar negatif

    Perspektif dot product :
    - Bentuk aljabar (Aturan cosinus)
    
        $$a \cdot b = a_1b_1 + a_2b_2 + ... + a_nb_n$$
    - Bentuk geometris
    
        $$ a \cdot b = ||a|| \; ||b|| \; \cos(\theta) $$
        
    Panjang proyeksi × panjang vektor lain

    $$
    \text{Cosine} \;\text{similarity} = \frac{a \cdot b}{||a|| \; ||b||}
    $$

    Cosine similarity mengukur : “Seberapa mirip arah dua vektor” 

    | Nilai Cosine | Nilai Theta |Makna                        |
    | ----- | ---  |---------------------------- |
    | 1      | 0 |Arah sama (identik)          |
    | 0     | 90 |Orthogonal (tidak berkaitan) |
    | -1    | 180 |Berlawanan arah              |

    **Fun Fact**
    - Nilai dot product antara vector yg saling tegak lurus = 0
    - Jika antar vectro saling tegak lurus maka nilai $\theta$ dari cosine adalah $90^\circ$ maka nilai dari cosine itu sendiri adalah 0
    - Nilai dot product dengan dirinya sendiri itu = 1
- Cross Product

#### **Faktor regangan**

$$
\text{Regangan} = \frac{||Av||}{||v||}
$$

Arah berbeda maka regangan pun berbeda, lalu muncul pertanyan <br>
Arah mana yang punya regangan “paling bersih / konsisten”?

Maka ada arah spesial yakni 
$$
A^TAv = \lambda v
$$

Ini spesial karena 
$$
\begin{align*}
||Av||^2 &= v^T A^T Av \\
&= \lambda ||v||^2 \\
||Av|| &= \sqrt{\lambda} ||v|| \\
&= \sigma ||v||
\end{align*}
$$

v yang hasil dari eigenvector $A^T A$ itu disebut right singular vector (ada juga left jadi A^T di kanan), lalu v ini adalah arah dimana matriks A bekerja bekerja secara "teratur" 
$$
||Av|| = \sigma ||v||
$$

Artinya norm hasil vector transformasi itu $\sigma$ kalinya norm vector awal yang dimana ini hanya berlaku untuk vector singular

lalu ada suatu vector yang searah dengan $Av$ tapi dia punya norm = 1, yakni
$$
\begin{align*}
u &= \frac{Av}{||Av||} \\
||u|| &= 1
\end{align*}
$$

**Dalam Ranah SVD**
- V

    Semua vector $v_i$ yang dimana dia itu vectro singular dikumpulkan dalam 1 matriks

    Kita menggunakan transposenya karena kita mau mentrasnformasi vector singular ke basis standar sedangkan vectro asli yg kita ingin trasnformasi mendapatkan "efek transformasinya saja"
- $\Sigma$

    Semua nilai singular yang menunjukkan rasio perbandingan norm vectro hasil trasnformasi terhadap vector yang ditrasnformasi dikumpulkan dalam 1 matriks

    $$
    \begin{align*}
    ||Av|| &= \sigma ||v|| \\
    \sigma &= \frac{||Av||}{||v||}
    \end{align*}
    $$
- U

    Semua vectro $u_i$ yang searah dengan $Av$ tapi dia punya norm = 1 dikumpulkan dalam 1 matriks

#### **Relasi Geometris**

- Ortonormal

    Dua vektor dikatakan ortogonal jika hasil dot product-nya = 0
    $$ v \cdot u = 0 $$

    Mereka saling tegak lurus misal ada yang ke atas dan ada yang ke kanan `(|__)`

    Mereka tidak punya “komponen arah yang sama” juga proyeksi satu ke yang lain = 0
- Ortogonal

    Vector dikatakan ortogonal saat ortonormal dan panjangnya = 1 (unit vector)
    $$ v \cdot u = 0, \quad ||v|| = 1 \;\text{dan}\; ||u|| = 1$$

#### **Proyeksi Vector**

Ada 2 vectro dan juga cahaya, saat suatu vectro x diberi cahaya dari arah A kita dicari dimana bayangan vector x yg jatuh tepat searah dengan vector v
- Secara Umum

    $$
    \begin{align*}
    \mathrm{proj}_{\mathbf{v}}(\mathbf{x}) &= \frac{\mathbf{x} \cdot \mathbf{v}}{\mathbf{v} \cdot \mathbf{v}} \, \mathbf{v} \\
    &= \left(\frac{\mathbf{x} \cdot \mathbf{v}}{\|\mathbf{v}\|}\right) \, \mathbf{u} \\
    &= \frac{\mathbf{x} \cdot \mathbf{v}}{\|\mathbf{v}\|^2} \, \mathbf{v} 
    \end{align*} 
    $$
    - $\mathbf{u} :$ searah dengan vector v tapi norm = 1 

        $$
        \mathbf{u} = \frac{\mathbf{v}}{\mathbf{||v||}}
        $$
    - $\mathbf{x} \cdot \mathbf{v}$ dan $\mathbf{v} \cdot \mathbf{v}$


        Ini perkalian dot product
    - $k = \frac{\mathbf{x} \cdot \mathbf{v}}{\mathbf{v} \cdot \mathbf{v}} = \frac{\mathbf{x} \cdot \mathbf{v}}{\|\mathbf{v}\|^2} $

        Ini akan menghasilkan scalar yg akan dikalian dengan vector v dikanan rumus
- Mekanisme saat bayangan dengan sudut siku-siku

## **Matriks**

Matriks adalah susunan angka dalam baris dan kolom : satu kolom matrisk bisa bermakna 1 vector, satu baris matriks bisa dari persamaan, dan bisa juga dari dataframe data, tiap kolom itu tiap variabel 

**1 Kolom adalah 1 garis vector lalu barisnya pertanda dimensi**

- Bentuk Matriks
    - Matriks segitiga
        - Atas

            $$ \begin{bmatrix} 1 & A & B \\ 0 & 1 & C \\ 0 & 0 & 1 \end{bmatrix} $$
        - Bawah
 
            $$ \begin{bmatrix} 1 & 0 & 0 \\ A & 1 & 0 \\ B & C & 1 \end{bmatrix} $$
    - Matriks diagonal 

        Matriks yang hanya memiliki elemen tak nol di diagonal utama
        
        $$ \begin{bmatrix} d_{11} & 0 & 0 \\ 0 & d_{22} & 0 \\ 0 & 0 & d_{33} \end{bmatrix} $$

        **Dalam trasnformasi linear ini bekerja dengan `meregangkan` setiap sumbu**
    - Matriks ortogonal  

        Kolom-kolom matriks ortogonal itu saling ortogonal dan panjangnya 1

        matriks ortogonal = kumpulan vektor ortonormal

        Syarat matriks ortogonal
        $$
        \begin{align*}
        Q^T Q &= I \\
        Q^T &= Q^{-1} 
        \end{align*}
        $$

        **Dalam trasnformasi linear dengan bekerja dengan `merotasikan atau merefleksikan` setiap sumbu, ini transformasi yang menjaga bentuk jaid tidak mengubah panjang vector dan sudut vector**
    - Matriks simetris

        matrisk persegi yang nilai entrynya indentik denga 2 sisi diagonal
        $$
        \begin{bmatrix}
        2 & (4) & [5] \\ (4) & 7 & <8> \\ [5] & <8> & 0
        \end{bmatrix}
        $$

        Ini memili sifat yang kuat, yakni vector eigennya saling tegak lurus `(|__)` <br>
        maka saat kita menormalkan vector eigen tersebut dan mengemasnya dalam matriks maka kita dapat matriks ortogonal (yang menyiratkan rotasi)

        **Ada kemampuan untuk membangun matriks simetris** <br>
        (Berlaku untuk semua A)
        $$A =  \begin{bmatrix} 1 & 2 & 3 \\ 4 & 5 & 6 \end{bmatrix} $$
        - $A A^T \rightarrow S_L \;(\text{Left Singular Vector})$

            $$\begin{bmatrix} 14 & 32 \\ 32 & 77 \end{bmatrix} $$
        - $A^T A \rightarrow S_R \;(\text{Right Singular Vector})$

            $$\begin{bmatrix} 17 & 22 & 27 \\ 22 & 29 & 36 \\ 27 & 36 & 45 \end{bmatrix} $$

        Fun Fact Kedua matriks di atas
        - Kedunya tergolong matriks PSD (Positive semi-definite)
        - Nilai eigen tidak akan negatif $(\lambda \geq 0)$
        - Semua eigenvalue yang tidak nol (non-zero) adalah sama antar matrisk $S_L$ dan $S_R$

            Singular value $(\sigma_i) = \sqrt{\lambda_i}$

Saat vector awal (V1) berada di dimensi awal lalu (V1) setelah dikali oleh matriks berada di dimensi yang bereda **(Dimensi tampak melebar atau menyempit)**

Namun, antara dimensi awal dan baru **kedua vector itu memiliki `jarak x dan y yang sama secara angka jarak`**, **Lihat pada garis biru dan merah keduanya memiliki `angka jarak yang sama` (Biru = 2, Merah = -1)**. Garis tersebut berubah setelah dikali oleh matriks bisa dilihat biru yang awalnya berdiri tegak lurus sekarang menjadi tiduran <br>
Saat di dimensi awal saat garis biru bergerak 2 langkah menurut vector [0, 1] itu berada di titik = 2 sedangkan dimensi baru garis biru bergerak 2 langkah itu menurut vector [3, 0] itu berada di titik = 6 <br>
**Satuan titik untuk vector berbeda biarpun jaraknya sama antar dimensi** 

Pergerak :
- Pada dimensi awal : biru bergerak ke atas 2 langkah jadi [0, 2] > lalu bergerak ke kiri 1 langkah dari hijau maka akan berada di titik [-1, 2]
- Pada dimensi baru : biru bergerak ke kanan [Awalnya atas] 2 langkah jadi [6, 0] > lalu bergerak ke atas 1 langkah dari hijau maka akan berada di titik [5, 2]

Untuk penentuan pergarak negatif dan positif sifatnya masih menggunakan dimensi awal yang dimana positif (kanan, atas) dan negatif (kiri, bawah). Di dimensi baru juga sama namu sesuikan garis jangan langsung  (kanan, atas, kiri, bawah)

### **Determinan $(|A|)$**

Determinan berbicara tentang luas

$$
\text{det} \left( 
\begin{bmatrix}
a & b \\
c & d \\
\end{bmatrix}
\right) = a(d) - b(c)
$$

Ada perkalian antar matriks 
$$
\begin{align*}
(A)(B) &= (C) \\
\begin{pmatrix}
2 & 0 \\
0 & 3 \\
\end{pmatrix}
\begin{pmatrix}
2 & 0 \\
0 & 2 \\
\end{pmatrix} &=
\begin{pmatrix}
4 & 0 \\
0 & 6 \\
\end{pmatrix}
\end{align*}
$$

Luas determinan 
- A : 6 
- B : 4 

kan ada perkalian antar matriks yg akan menghasilkan matriks baru, nah matriks baru yg dihasilkan itu harus memiliki determinan dari perkalian determinan matriks yg menghasilkannya

$$
\begin{align*}
\text{Determinan}\; C &= (\text{Determinan}\; A)(\text{Determinan}\; B) \\
24 &= (6)(4)
\end{align*}$$

### **Inverse $(A^{-1})$**

$$
\begin{align*} 
A^{-1} &= \frac{1}{\text{det}(A)} \text{adj}(A) \\
&= \frac{1}{ad - bc} \begin{pmatrix} d & -b \\ -c & a \end{pmatrix} \\
\text{adj}(A) &= \text{det}(A) A^{-1}
\end{align*} 
$$

**Sifat invers matriks**
- $(A^{-1})^{-1} = A$
- $(AB)^{-1} = B^{-1} A^{-1}$
- $AA^{-1} = A^{-1}A = I$ 
- $|A| = \frac{1}{|A^{-1}|}$
- $\lambda_{A^{-1}} = \frac{1}{\lambda_{A}}$  

Cara mencari adjoin matriks 3x3 dan seterusnya
$$
\begin{align*}
A &= 
\begin{bmatrix}
a & b & c \\ d & e & f \\ g & h & i
\end{bmatrix} \\
C &=
\begin{bmatrix}
+(ei-fh) & -(di-fg) & +(dh-eg) \\ -(bi-ch) & +(ai-cg) & -(ah-bg) \\ +(bf-ce) & -(af-cd) & +(ae-bd)
\end{bmatrix} \\
\text{adj}(A) &= C^T
\end{align*}
$$

#### **Singular Value Decomposition (SVD)**

**Refrensi :** 
- https://youtu.be/llisH02KLrE?si=g5zwDD2Pjrde87N6
- https://youtu.be/vSczTbgc8Rc?si=zJkPO8tpw1Djiwbf

$$ 
\begin{align*}
A &= U \Sigma V^T \\
\end{align*}
$$
- $U :$ Eigen vector dari matriks $A A^T$
- $\Sigma :$ Nilai singular (Akar dari Eigen Value $A A^T$ atau $A^T A$)
- $V^T :$ Eigen vector dari matriks $A^T A$

Transformasi dilakukan dari kanan ke kiri
- $V^T$ : Matriks $A^T A$ (Right singular matriks) yang ditranspose

    **Transformasi ini yang membuat vector singular (mereka ortonormal) hasil dari nilai eigen matriks singular kembali ke basis standar**

    Matriks $V^T$ berisi vektor-vektor singular tersebut sebagai baris-barisnya.
    $$V^T = \begin{bmatrix} \leftarrow & v_1 & \rightarrow \\ \leftarrow & v_2 & \rightarrow \\ \leftarrow & v_3 & \rightarrow \end{bmatrix}$$
    
    Saat kita melakukan transformasi terhadap salah satu vektor singular, misalnya $v_1$, maka secara matematis terjadi perkalian baris dikali kolom:
    $$V^T v_1 = \begin{bmatrix} v_1 \cdot v_1 \\ v_2 \cdot v_1 \\ v_3 \cdot v_1 \end{bmatrix}$$
    
    Berdasarkan hukum ortonormal tadi:
    - Baris 1 ($v_1 \cdot v_1$) menghasilkan 1.
    - Baris 2 ($v_2 \cdot v_1$) menghasilkan 0 (karena mereka tegak lurus).
    - Baris 3 ($v_3 \cdot v_1$) menghasilkan 0 (karena mereka tegak lurus).

    Hasil akhirnya adalah $\begin{bmatrix} 1 \\ 0 \\ 0 \end{bmatrix}$ 
    
    Inilah cara ajaib bagaimana arah yang miring "diluruskan" menjadi sumbu standar. 
    
    Yang lainnya juga akan berakhir mirip dapat "0" dapat 2 lalu "1" dapat 1 karena ada keadaan dimana dia akan dikali oleh dirinya sendiri dan menjadi 1

    Menjajajarkan vektor dengan "Sumbu Utama" (Basis Standar). Matriks $V^T$ mencari tahu koordinat baru yang paling efisien untuk melakukan regangan.

    “vector asli akan mendapatkan efeknya saja” yg dimana efek itu adalah efek bagaimana vectro singular ditrasnformasi ke basis standar
- $\Sigma$ : Matriks berisi nilai singular dari ujung kiri atas ke kanan bawah, lalu jika ini matrisk persegi panjang maka paling kanan atau paling bawah itu berisi nilai 0

    Ini adalah nilai $\sqrt{\lambda}$ dari matriks $A^T A$ atau $A A^T$ (keduanya punya nilai eigen sama) **`[Urutkan dari yang terbesar dari ujung kiri atas ke kanan bawah]`**

    **Transformasi ini yang mengatur dimensi agar menuju dimensi yang telah ditentukan, disini juga yang membuat dimensi meregang sesuai nilai singular (Setiap Maju 1 langkah maka akan menghasilkan nilai singular, bukan seperti biasa maju 1 langkah menghasilkan 1)**

    **Tempat penentuan apakah akan menghapus dimensi atau menambah dimensi**
- $U$ : Matriks $A A^T$ (Left singular matriks)

    **Transformasi ini yang memutar matriks basis standar agar sejajar dengan matriks singular kiri**

$$
\begin{align*}
\begin{bmatrix}
3 & 2 & 2 \\
2 & 3 & -2
\end{bmatrix} &= 
\left[
\begin{bmatrix}
3 & 2 & 2 \\
2 & 3 & -2
\end{bmatrix}
\begin{bmatrix}
3 & 2 \\ 2 & 3 \\ 2 & -2
\end{bmatrix}
\right]
\begin{bmatrix}
\sigma_1 & 0 & 0 \\ 0 & \sigma_2 & 0
\end{bmatrix}
\left[
\begin{bmatrix}
3 & 2 \\ 2 & 3 \\ 2 & -2
\end{bmatrix}
\begin{bmatrix}
3 & 2 & 2 \\
2 & 3 & -2
\end{bmatrix}
\right]^T \\
&= \begin{bmatrix}
\frac{1}{\sqrt{2}} & -\frac{1}{\sqrt{2}} \\
\frac{1}{\sqrt{2}} & \frac{1}{\sqrt{2}}
\end{bmatrix}
\begin{bmatrix}
5 & 0 & 0 \\
0 & 3 & 0
\end{bmatrix}
\begin{bmatrix}
\frac{1}{\sqrt{2}} & \frac{1}{\sqrt{2}} & 0 \\
-\frac{1}{\sqrt{18}} & \frac{1}{\sqrt{18}} & -\frac{4}{\sqrt{18}} \\
\frac{2}{3} & -\frac{2}{3} & -\frac{1}{3}
\end{bmatrix}
\end{align*}
$$

Mekanisme Bagaimana Munculnya Konsep SVD
- Ada suatu matriks

    $$ A = \begin{bmatrix} 3 & 2 & 2 \\ 2 & 3 & -2 \end{bmatrix}    
- Kitau tahu bahwa matriks tersebut gunanya untuk mentransformasi suatu vector dari suatu dimensi ke dimensi lain
- Kita ambil 2 vector yg saling ortogonal [Tegak Lurus] `(|__)`

    Kemudian kita **cari nilai `2 vector`** tersebut (Secara visuali di putar) yang **saat di `transformasi` oleh `matriks`** menghasilkan **vector `baru` saling `tegak lurus`** (nilai dot productnya maka 0)

$$
\begin{align*}
1. \quad A v_1 &= y_1 \\
&= \sigma_1 u_1 \\
&= ||\;y_1\;|| u_1 \\
2. \quad A v_2 &= y_2 \\
&= \sigma_2 u_2 \\
&= ||\;y_2\;|| u_2
\end{align*}
$$
- Kemudian cari nilai norm (panjang vector) untuk setiap vector hasil transformasi $\left(||\;y_i\;|| \right)$ yang mana itu akan menjadi nilai $\sigma_i$

    Dengan itu untuk mencari nilai $u$ dengan cara dia sebagai pembagi nilai setiap nilai vector hasil transformasi

    $$ 
    \begin{align*}
    u_i &= \frac{y_i}{||y_i||} \\
    &= \frac{y_i}{\sigma_i} 
    \end{align*}
    $$ 
- Membentuk formula SVD 
    $$A \underbrace{[v_1 \quad v_2]}_{V} = \underbrace{[u_1 \quad u_2]}_{U} \underbrace{\begin{bmatrix} \sigma_1 & 0 \\ 0 & \sigma_2 \end{bmatrix}}_{\Sigma}$$
    - $V$ : Vector yg ingin di trasnformasi dengan matriks jadikan 1 menjadi matriks 
    - $U$ : Kumpulan Vector hasil transformasi yg telah dibagi oleh nilai singular
    - $\Sigma$ : leading onenya berisi nilai singular, disini dalam matriks m x n dapat berperan sebagai penambah atau penghapus dimensi

    $$ 
    \begin{align*}
    AV V^{-1} &= U \Sigma V^{-1} \\ 
    A I &= U \Sigma V^T \\
    A &= U \Sigma V^T 
    \end{align*}$$

    Karena V adalah matriks ortogonal, maka nilai $V^{-1} = V^T$


Awalnya kan kita punya vector awal yg ditransformasi oleh matriks A untuk membentuk vector baru, kita rotasi vector awal tersebut maka dia akan membentuk bermacam vector baru. Kolom pertama matriks v (bukan $V^T$ jadi jika di hasil dia berupa baris) adalah vector awal yang jika ditranformasi oleh matriks A maka akan menghasilkan vector hasil trasnformasi yang memiliki norm terbesar daripada berbagai macam vector baru hasil dari vectro awal dirotasi.  

$v_1$ punya panjang tertentu lalu setelah ditransformasi oleh matriks maka panjang mengikuti nilai singular 

Kolom pertama matriks U adalah vector yg searah dengan kolom pertama matriks V yang ditranformasi oleh matriks A (Dia punya norm paling besar) tapi bedanya matriks U kolom pertama punya norm = 1 <br>
Juga kolom pertama matriks U ini tempat mendaratnya dari kolom pertama matriks V yg ditrasnformais oleh matriks A
$$A v_i = \sigma_i u_i$$

##### **Python**

```r
import numpy as np

# Definisi matriks
A = np.array([
    [3, 1],
    [1, 3],
    [1, 1]
])
print(f"Matrix A:\n{A}")
```

```r
# SVD
U, S, VT = np.linalg.svd(A)
print(f"\nMatrix U:\n{U}")
print(f"\nSingular Values (S):\n{S}")
print(f"\nMatrix V^T:\n{VT}")

# Membentuk Sigma matrix
Sigma = np.zeros((A.shape[0], A.shape[1]))
np.fill_diagonal(Sigma, S)

print(f"\nMatrix Sigma:\n{Sigma}")
```

```r
# Verifikasi, kalikan semua komponen untuk mendapatkan kembali A
A_reconstructed = U @ Sigma @ VT
print(f"Reconstructed A:\n{A_reconstructed}")
```

$$
X_k = U_k \Sigma_k V_k^T
$$
- `U[:, :k]` -> ambil k kolom pertama dari $U$
- `S[:k]` -> ambil k singular value terbesar
- `Vt[:k, :]` -> ambil k baris pertama dari $V^T$

k menentukan berapa banyak “blok rank-1” yang dipakai untuk menyusun matriks

```r
# daftar k yang ingin ditampilkan
k_values = [100, 80, 60, 40, 20, 10]

plt.figure(figsize=(12, 8))

for i, k in enumerate(k_values):
    # rekonstruksi gambar dengan rank-k
    # compressed_image = np.dot(U[:, :k], np.dot(np.diag(S[:k]), Vt[:k, :]))
    compressed_image = U[:, :k] @ np.diag(S[:k]) @ Vt[:k, :]
    
    # subplot 2 baris 3 kolom
    plt.subplot(2, 3, i+1)
    plt.imshow(compressed_image, cmap='gray')
    plt.title(f'k = {k}')
    plt.axis('off')

plt.tight_layout()
plt.show()
```

```r
# Save file kompres
k = 50
compressed_image = U[:, :k] @ np.diag(S[:k]) @ Vt[:k, :]
plt.imshow(compressed_image)
plt.savefig("image/burung_50.png")
```

```r
# load gambar
image2 = io.imread('image/burung.jpeg')

# siapkan array hasil
reconstructed = np.zeros_like(image2, dtype=float)

# SVD per channel karena ini bukan warna grey jadi ada banyak warna
for i in range(3):  # R, G, B
    U, S, Vt = np.linalg.svd(image2[:, :, i], full_matrices=False)
    reconstructed[:, :, i] = (U * S) @ Vt

# clipping agar valid (0–255)
reconstructed = np.clip(reconstructed, 0, 255).astype(np.uint8)

# tampilkan
plt.imshow(reconstructed)
plt.axis('off')

# simpan
plt.savefig("image/burung_rekonstruksi_full.png", bbox_inches='tight', pad_inches=0)
plt.show()
```

#### **Diagonalisasi Matriks**

Ini berguna saat perkalian matriks yg berpangkat (jadi terus mengali dirinya sendiri dengan perkalian matriks) 

$$
A = P D P^{-1}
$$
- $P$ : Berisi vector eigen yang tempat disesuaikan dengan tempat nilai eigen, jika nilai eigennya di matriks $D$ berada di pojok kiri atas maka vector eigennya mengisi kolom 1 dari kiri di matriks $P$ 
- $D$ : Diagonal utamanya berisi nilai eigenvalue yang dapat di susun berurutan dari yg terbesar dari kiri atas ke kanan bawah
- $P^{-1}$ : Versi invers dari matriks $P$

$$
\begin{align*}
A^n &= \begin{pmatrix} 1 & 0 \\ 6 & -1 \end{pmatrix}^n \\
&= (PDP^{-1})^n \\
&= \begin{pmatrix} 1 & 0 \\ 3 & 1 \end{pmatrix} \begin{pmatrix} 1^n & 0 \\ 0 & -1^n \end{pmatrix} \begin{pmatrix} 1 & 0 \\ -3 & 1 \end{pmatrix}
\end{align*}
$$

Jadi kita tidak perlu melakukan perkalian matriks berulangan saat ingin memangkatkannya, cukup memangkatkan matriks $D$ yang berisi eigenvalue lalu dilanjutkan proses perkaliannya

```r
B <- matrix(c(4,2,5,4,8,6,7,3,9), nrow = 3, ncol = 3, byrow = TRUE)
print(B)
```

```r
eig <- eigen(B)

P <- eig$vectors        # eigenvectors
print(P)
lambda <- eig$values    # eigenvalues

# Matriks diagonal
D <- diag(lambda)
print(D)

# Invers matriks P
P_inv <- solve(P)
print(P_inv)
```

```r
# Rekonstruksi
dekom_rekon = P %*% D %*% P_inv
print(dekom_rekon)
```

#### **QR Decomposition**

Metode ini menggunakan rotasi bidang untuk mengnolkan elemen tertentu dalam matriks secara spesifik

Sangat berguna untuk matriks yang memiliki banyak elemen nol (sparse matrices) karena kita dapat memilih elemen mana yang ingin dieliminasi tanpa memengaruhi banyak elemen lainnya

$$
A = QR
$$

```r
B <- matrix(c(4,2,5,4,8,6,7,3,9), nrow = 3, ncol = 3, byrow = TRUE)
print(B)
```

```r
qr_B <- qr(B)
qr_B$qr
qr_B$qraux

Q <- qr.Q(qr_B)
Q
R <- qr.R(qr_B)
R
```

```r
# Verifikasi dengan merekonstruksi matriks B
B_qr <- Q %*% R
print(B_qr)
```
