##### **Other**

- `\begin{cases} \end{cases}`

    $$
    f(x) =
    \begin{cases}
    x^2 & x > 0 \\
    0 & x \le 0
    \end{cases}
    $$
- `\begin{align} \end{align}` 

    Yang membuat sejajar adalah `&`

    $$
    \begin{align}
    y &= mx + c \\
    &= 2x + 3 \\
    &= 2(5) + 3 \\
    &= 13
    \end{align}
    $$
    
    Jika tidak ingin ada angka di kanan, gunakan `align*`
    $$
    \begin{align*}
    a &= b + c & d &= e + f \\
    x &= y - z & p &= q + r
    \end{align*}
    $$
- `\begin{gather} \end{gather}` 

    Mirip dengan align ini versi dikumpulkan di tengah
    $$
    \begin{gather}
    x + y = 10 \\
    x - y = 4
    \end{gather}
    $$

    Jika tidak ingin ada angka di kanan, gunakan `gather*`
