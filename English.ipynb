### **Aktif vs Pasif**

- `Continuous`    

    Digunakan ketika kita melihat suatu tindakan sebagai proses yang sedang berlangsung pada suatu titik waktu.
    - Aktif
    
        $$\boxed{\text{S + to be + v-ing}}$$

        `I am washing my car` = `Aku sedang mencuci mobilku`

        Perluasan konteks: Aku sedang berada dalam proses mencuci mobil. Kegiatannya belum dilihat sebagai sesuatu yang selesai, tetapi sebagai aktivitas yang sedang berlangsung
    - Pasif
    
        $$\boxed{\text{S + to be + being + v3}}$$

        `My car is being washed` = `Mobilku sedang dicuci`

        Mobilku sedang berada dalam proses pencucian. Kita tidak perlu menyebutkan siapa yang mencucinya.
- `Perfect`

    Ada sesuatu yang telah terjadi sebelum sekarang, dan hasil atau dampaknya masih relevan sekarang.
    - Aktif

        $$\boxed{\text{S + have/has/had + v3}}$$

        Melihat dari pelaku, cara berpikirnya adalah dengan memahami siapa yang melakukan tindakan itu?

        `I have washed my car` = `Aku sudah menyuci mobilku`

        Subjek `(I)` yang bertindak sebagai pencuci mobil

        Perluasan konteks: Aku melakukan kegiatan menyuci mobil sebelumnya, dan hasilnya masih relevan sekarang (misalnya mobilku jadi terlihat bersih)

        Dalam contoh di atas, yang ingin disampaikan adalah subjek `(I)` melakukan sesuatu `(washed)` lalu `(my car)` hanyalah objek
    - Pasif
    
        $$\boxed{\text{S + have/has/had + been + v3}}$$

        Melihat dari objek
        - Present

            `My car has been washed` = `Mobilku sudah dicuci`

            Subjek bukan yang bertindak sebagai pencuci mobil

            Fokusnya bukan siapa yang menyelesaikan, fokusnya adalah mobil sekarang berada dalam keadaan sudah selesai dicuci.
        - Past

            `My car had been washed before I left` = `Mobil saya sudah dicuci sebelum saya berangkat`

            $$\text{Masa depan --- Sekarang --- (Berangkat) --- (Mobil sudah dicuci)}$$

            `when i went to home, food had already been prepared`

            Konteksnya dapat dipahami saat bercerita (cerita pengalaman atau kejadian dimasa lalu)
- `Perfect Continuous`

    Digunakan ketika kita melihat suatu tindakan sebagai proses yang sudah berlangsung selama suatu periode waktu sampai titik acuan tertentu.
    - Aktif

        $$\boxed{\text{S + have/has/had + been + v-ing}}$$

        Melihat dari pelaku dan menekankan proses atau durasi tindakan.

        `I have been washing my car for two hours` = `Saya sudah mencuci mobil saya selama dua jam`

        Aku mulai melakukan aktivitas tersebut sebelumnya → aktivitas berlangsung selama suatu periode → sekarang aktivitas itu masih berlangsung atau baru saja selesai dan hasil/prosesnya masih relevan.

        - `perfect` $\to$ sudah terjadi kejadian
        - `perfect continuous` $\to$ sudah berlangsungnya kejadian
    - Pasif

        Tidak umum digunakan karena ada bentuk yg lebih sederhana

## **Conditional Sentences**

| Tipe       | If Clause    | Main Clause                       | Makna                                                                 |
| ---------- | ------------ | --------------------------------- | --------------------------------------------------------------------- |
| **Type 1** | **V1**       | **will/can/may + V1**             | Kondisi nyata atau masih mungkin terjadi.                             |
| **Type 2** | **V2**       | **would/could/might + V1**        | Kondisi tidak nyata atau bertentangan dengan keadaan sekarang.        |
| **Type 3** | **had + V3** | **would/could/might + have + V3** | Kondisi masa lalu yang tidak terjadi dan hasilnya juga tidak terjadi. Dapat dipahami juga dengan nuansa penyesalan |

### **Zero**

Fakta umum, kebenaran ilmiah, kebiasaan, sesuatu yang selalu benar, ...

$$ \text{If + Simple Present, Simple Present} $$
- If you heate water to $100^{\circ}$ C, it boils = Jika memanaskan air, air mendidih

    Ini jelas merupakan fakta ilmiah
- `If I drink coffee at night, I can't sleep` = `Kalau minum kopi malam hari, aku tidak bisa tidur`

    Ini kebiasaan
- `If people don't eat, they die`

    Ini fakta umum

### **First**

Kemungkinan dimasa depan jika syarat terpenuhi

Situasi realistis dan masih mungkin terjadi
$$ \boxed{\text{If + S + v1, S + (will/can/may) + v1}} $$

`If it train tomorrow, we will cancle the trip`

`If you study hard, you will pass the test`

### **Second**

Situasi tidak relistis atau nyata untuk dimasa depan ataupun sekarang

Ini dalam konteks khayalan, situasi yang tidak nyata sekarang, kemungkinan yang sangat kecil, ..

$$ \boxed{\text{If + S + v2, (would/could/might) + v1}} $$

`If i won the lottery, i would travel the world`

Berhayal untuk menang lotre yang dimana itu tidak dapat terjadi di dunia nyata atau kemungkinannya sangat kecil

$$ \boxed{\text{S + (would/could/might) + v1 + if + S + v2}} $$

`I could swim if I had more time` <br>
= `Seandainya saya punya lebih banyak waktu (padahal sekarang tidak), saya bisa berenang`

### **Third**

Situasi tidak nyata dimasa lalu, penyesalan, kejadian masa lalu yang tidak bisa diubah, ..

Membicarakan keadaan masa lalu yang tidak terjadi.

$$ \boxed{\text{if + S + had + v3, S + (would/could/might) + have + v3}} $$

`If I had studied, I would have passed the test` = `Kalau dulu aku belajar, aku pasti lulus`

Kenyataannya tidak lulus karena tidak rajin belajar, lalu berandai dengan nuansa penyesalan dulu rajin belajar agar bisa lulus

$$ \boxed{\text{S + (would/could/might) + have + v3 + if + S + had + v3}} $$

`If I had studied harder, I would have passed the exam` = `Seandainya dulu saya belajar lebih giat, saya pasti sudah lulus ujian`

### **Mixed**

#### **Past $\to$ Present**

$$ \boxed{\text{If + had + V3, would + V1}} $$

**Masa lalu memengaruhi keadaan sekarang**

Sekilas mirip Second Conditional, bedanya ada pada maknanya. 

Main clause memakai would + V1 tetapi maksudnya keadaan sekarang.

`If I had studied harder, I would have a better job now` <br>
= `Kalau dulu saya belajar lebih giat, sekarang saya akan memiliki pekerjaan yang lebih baik`

#### **Present → Past**

$$ \boxed{\text{If + V2, would have + V3}} $$

**Keadaan sekarang memengaruhi kejadian masa lalu**

If clause memakai Second Conditional

Main clause memakai Third Conditional

`If I were taller, I would have become a basketball player` 
- `Jika saya lebih tinggi, saya pasti sudah menjadi pemain bola basket` (trasnlate)
- `Kalau aku memang orang yang tinggi, mungkin dulu aku sudah menjadi pemain basket` (chatgpt)

Karena sekarang aku pendek maka dulu aku tidak menjadi pemain basket

`If I were more organized, I wouldn't have forgotten the meeting yesterday` 
- `Seandainya saya lebih teratur, saya tidak akan lupa rapat kemarin` (trasnlate)
- `Kalau aku memang orang yang lebih teratur, kemarin aku tidak akan lupa rapat` (chatgpt)

Keadaan sekarang memang aku orang ynag ceroboh, akibatnya kemarin aku lupa rapat

#### **Perbandingan**

Ketiga jenis conditional ini sama-sama membicarakan situasi yang tidak nyata (unreal condition). Namun, yang membedakan adalah waktu kondisi (if clause) dan waktu hasilnya (main clause)

| Jenis                       | Pola                              | Kalimat                                                      | Arti                                                                                      |
| --------------------------- | --------------------------------- | ------------------------------------------------------------ | ----------------------------------------------------------------------------------------- |
| **Third Conditional**       | `If + had + V3, would have + V3`  | `If I had studied harder, I would have passed the exam.`     | Jika dulu saya belajar lebih giat, saya pasti sudah lulus ujian.                          |
| **Mixed<br>Past → Present** | `If + had + V3, would + V1`       | `If I had studied harder, I would have a better career now.` | Jika dulu saya belajar lebih giat, sekarang saya akan memiliki karier yang lebih baik.    |
| **Mixed<br>Present → Past** | `If + V2 (were), would have + V3` | `If I were more disciplined, I would have studied harder.`   | Jika saya adalah orang yang lebih disiplin (sekarang), dulu saya akan belajar lebih giat. |

Pemahaman lanjut
- **Third Conditional**

    "Dulu begini, dulu hasilnya juga akan berbeda"
    
    Kondisinya tidak terjadi di masa lalu. Akibatnya, hasilnya juga tidak terjadi di masa lalu

    Artinya sekarang semuanya sudah selesai

    Fokusnya adalah: "Apa yang seharusnya terjadi dulu"
- **Mixed Conditional (Past → Present)**

    "Dulu begini, maka hidup saya sekarang akan berbeda"

    Yang dibahas sebenarnya bukan masa lalunya.

    Yang dibahas adalah: "keadaan saya sekarang"

    Masa lalu hanya dijadikan penyebab.
- **Mixed Conditional (Present → Past)**

    "Karena saya sekarang seperti ini, dulu hasilnya menjadi seperti itu"

    Berbeda dengan dua sebelumnya.

    Di sini penyebabnya adalah: karakter atau kondisi sekarang yang dianggap sudah lama ada.

    Fokus utamanya bukan kejadian masa lalu. Tetapi: "siapa saya sekarang"

    `If I were more confident, I would have applied for that job` <br>
    = `Jika saya lebih percaya diri, saya pasti sudah melamar pekerjaan itu`

    Makna lainnya saya sekarang bukan orang yang percaya diri, karena itulah dulu saya tidak melamar pekerjaan itu.

## **Pemahaman Advanced**

#### **`Would` | `Will`**

`How big would Eren's Founding Titan be if it formed correctly?` <br> 
= `Seberapa besar Founding Titan milik Eren jika ia terbentuk dengan benar?`

kenyataannya Titan tersebut tidak terbentuk secara normal lalu kita membayangkan kondisi alternatif $\to$ hypothetical situation

- Mengapa pakai would?

    Would digunakan untuk menyatakan sesuatu yang merupakan dugaan atau keadaan yang tidak nyata.

    Jadi would menunjukkan "apa yang akan terjadi dalam dunia hipotesa"
- Bagaimana jika menggunakan will

    **How big will Eren's Founding Titan be if it forms correctly?**

    "Seberapa besar Titan Eren nanti jika ia benar-benar terbentuk dengan benar?"

    Di sini penuturnya percaya bahwa proses pembentukan masih mungkin terjadi di masa depan

    - `would`: ini berandai di dunia hipotesa yang mana di kenyataannya tidak terjadi
    - `will`: Belum terjadi dan kita berandai di dunai nyata kedepannya
- Kalau menggunakan is

    How big is Eren's Founding Titan? = "Berapa ukuran Titan Eren?"

    Ini menanyakan ukuran yang benar-benar ada sekarang, bukan ukuran alternatif

##### **`...Would be...`**

A: `Maybe we should study together tomorrow`

B: `That would be a good idea`

Menggunakan `would` disini karena memberikan respon yang lebih sopan, lebih tidak langsung (less direct), dan lebih hipotetis

Jika kita menggunakan `will` maka nuansanya akan sudah pasti menjadi ide bagus

`If I had more money, I would be happy` = `Kalau aku punya lebih banyak uang, aku akan bahagia`

Hipotesis saat memiliki uang, sekarang belum punya

`Life would be boring without music` = `Hidup akan membosankan tanpa musik`

#### **`Do` | `Did`**

Auxiliary adalah kata kerja yang membantu kata kerja utama (main verb) untuk menyampaikan informasi tambahan.

- do → pembicara melihat kejadian sebagai sekarang, umum, atau belum selesai

    Menanyakan keadaan sekarang, kebiasaan, atau fakta.
    - Do you like coffee? = Apakah kamu suka kopi?

        Artinya: Sekarang atau secara umum, apakah kamu menyukai kopi?
- did → pembicara melihat kejadian sebagai masa lalu atau sudah selesai

    Menanyakan kejadian yang sudah lewat.
    - Did you like the movie? = Apakah kamu menyukai filmnya?

        Artinya: Filmnya sudah ditonton, sekarang aku bertanya bagaimana pendapatmu waktu itu.
    - Did you work here? = Apakah dulu kamu bekerja di sini?

        Fokusnya bukan pekerjaan sekarang, melainkan pekerjaan yang sudah selesai.
    - Penekanan

        I did call you = Aku benar-benar meneleponmu

#### **`Have` | `Had`**

##### **`+ Been`**

Penggunaannya sebagai
- Kata kerja utama (main verb) → bermakna memiliki, mengalami, menerima, melakukan, ...
- Auxiliary (kata kerja bantu) → membentuk Perfect Tense. Pada fungsi ini, maknanya berubah menjadi konsep hubungan waktu, bukan lagi "memiliki"

Penjelasan
- Have
    - Tanpa Been

        $$ \text{Have + v3} $$
        - `Have you eaten` = `Kamu sudah makan?`

            Yang ingin diketahui adalah Kondisimu sekarang bagaimana?
        - `Have you showered?` = `Apakah anda sudah mandi?`

            Fokusnya adalah hasil atau status saat ini, bukan proses mandinya

            `Are you showering?` = `Apakah kamu sedang mandi?`

            Fokusnya adalah proses yang sedang berlangsung, ingin tahu apakah saat ini sedang berada di kamar mandi
    - Dengan Been

        `Have you been...` = `(Apakah kamu sudah berada dalam suatu keadaan sampai sekarang?)`

        Menayakan kabar <br>
        `How have you been?` = `Apa kabar?`

        `...it has ever been` = `...yang pernah ada`

        `The economy is stronger than it has ever been` = `Perekonomian saat ini lebih kuat daripada sebelumnya`

        Penggunana `than` disitu sebagai perbandingan
        - Tambahan `v-ing` (Present Perfect Continuous)

            $$ \text{Have + been + v-ing} $$
            Sudah melakukan suatu sktivitas dan masih berlangsung sekarang, atau baru saja berhenti yang efeknya masih terasa
            - `I've been studying` = `Saya sudah belajar`

                Kegiatan belajar itu sudah selesai
                
                Fokus pada proses atau aktivitas belajar itu sendiri, nuansan yang diberikan bukan hanya selesai belajar saja melainkan selesai melakukan kegiatan belajar

                `I have studied` 
                
                fokus pada hasil atau fakta bahwa kegiatan belajar sudah terjadi

                **Perbedaan pertanyaan**
                - `What are you doing?` = `Kamu sedang apa?`

                    Fokusnya adalah momen sekarang.
                - `What have you been doing?` = `Apa yang selama ini Anda lakukan?`

                    Fokusnya adalah keadaan yang dilakukan dari waktu lalu hingga sekarang
        - Tambah `v3` (present perfect passive)

            $$ \text{Have/Has + been + v3} $$

            Maknanya seperti telah di..

            `The door has been opened` = `Pintu itu telah dibuka`

            `has the door been opened?` = `Apakah pintunya telah dibuka?`

            `The report has been finished` = `Laporan tersebut telah selesai`

            `You have looked happy lately` = `Belakangan ini kamu terlihat bahagia`

            `I have been studied by several researchers for my rare genetic condition` <br>
            = `Saya telah diteliti oleh beberapa peneliti karena kondisi genetik langka saya`
        - Lainnya

            $$ \text{Have + been + adjective/noun/prepositional phrase/adverb/to} $$
            - adjective (kata sifat)

                Menunjukkan keadaan yang berlangsung sampai sekarang

                Telah berada dalam suatu keadaan sampai sekarang

                `I have been busy` = `Aku sedang sibuk` (dan sampai sekarang masih sibuk)

                Bukan berarti "sedang menyibukkan diri", tetapi keadaan sibuk masih berlangsung.

                `We have been happy lately` = `Kami merasa bahagia belakangan ini`
            - noun (kata benda)

                Biasanya menunjukkan identitas, profesi, atau status

                Dapat dipahami seperti makna `telah menjadi`

                `I have been a teacher for ten years` = `Saya sudah menjadi guru selama sepuluh tahun`

                Artinya mulai menjadi guru 10 tahun lalu masih menjadi guru sekarang

                `She has been my best friend for years` = `Dia telah menjadi sahabat saya selama bertahun-tahun`
            - prepositional phrase 

                Menunjukkan keberadaan atau lokasi.

                `I have been at home all day` = `Saya berada di rumah seharian`

            - adverb

                Menjelaskan keadaan yang terjadi berkali-kali atau belakangan ini

                `Things have been better lately` = `Belakangan ini keadaan sudah lebih baik`

                `It has been difficult` = `Hal ini terasa sulit`
            - to

                pernah pergi ke... lalu sudah kembali

                `I have been to Bali` = `Aku pernah ke Bali`

                `Have you been to boston?` = `Apakah anda pernah ke boston?`
- Had
    - Tanpa Been

        $$ \text{Had + v3} $$

        Digunakan untuk menunjukkan bahwa suatu kejadian sudah selesai sebelum kejadian lain di masa lalu
        - `Yesterday, before the exam, had you eaten?` = `Kemarin, sebelum ujian, apakah kamu sudah makan?`

            Yang dicek adalah kondisi sebelum ujian kemarin, bukan sekarang 
        - `I had finished my homework before my mother came home` = `Aku sudah menyelesaikan PR sebelum ibu pulang`
    - Dengan Been
        - `+ v-ing`

            $$ \text{Had + been + v-ing} $$

            Digunakan untuk menunjukkan bahwa suatu aktivitas sedang berlangsung selama beberapa waktu sebelum titik tertentu di masa lalu
            - `I had been studying for three hours before the exam started` = `Aku telah belajar selama tiga jam sebelum ujian dimulai`
            - `She had been waiting for an hour before the bus arrived` = `Dia sudah menunggu selama satu jam sebelum bus datang`
        - `+ Noun`

            Menunjukkan status sebelum titik waktu di masa lalu.

            `He had been our manager before he retired` = `Dia pernah menjadi manajer kami sebelum pensiun.`
- Will + (Have/Has) + Been

    **Tidak bisa menggunakan `Had`**

    Menjelaskan keadaan yang akan sudah terjadi sebelum waktu tertentu di masa depan.

    `Next year I will (have been) a teacher for ten years` = `Tahun depan saya (sudah menjadi) guru selama sepuluh tahun`

    Yang ingin disampaikan pembicara adalah: Ketika tahun depan tiba, saya sudah mencapai 10 tahun sebagai guru.

    Mengapa memakai "been"? <br>
    Karena teacher adalah status, bukan aktivitas.

    `Next week we will (have been) friends for fifteen years` <br>
    = `Minggu depan, kami (sudah akan berteman) selama lima belas tahun`

    - `+ v3`

        `Next month they will (have been married) for ten years` <br>
        = `Bulan depan mereka (sudah menikah) selama sepuluh tahun`

        `The report will (have been finished) by tomorrow` = `Laporan itu akan (sudah selesai) pada hari esok`
    - `+ v-ing`

        Menunjukkan aktivitas yang akan sudah berlangsung selama periode tertentu pada waktu di masa depan

        `By next month I will have been working here for five years` <br>
        = `Bulan depan, saya akan sudah bekerja di sini selama lima tahun`

##### **`Would + have/has + been`**

Ini adalah pernyataan penyesalan atau evaluasi terhadap masa lalu yang tidak terjadi — semacam "seandainya saja..." Sesuatu yang lebih baik dilakukan, tapi sudah terlambat

`Taking a taxi would have been a good idea` = 
- `Naik taksi sebenarnya adalah ide yang bagus` 
- `Naik taksi akan menjadi ide yang bagus` 
- `Naik taksi mungkin merupakan ide yang bagus`

(seandainya dilakukan, tetapi kenyataannya tidak dilakukan)

Konsep lainnya
- Menghilangkan `been` dan menambahkan `had`

    `We ended up walking and got stuck in the rain. Taking a taxi would have had some advantages, like saving time and staying dry` <br>=`Kami akhirnya jalan kaki dan kehujanan. Naik taksi tadinya akan memberi beberapa keuntungan, seperti menghemat waktu dan tetap kering`

    Menyebutkan konsekuensi/manfaat konkret yang akan diperoleh, bukan sekadar menilai idenya

    Di sini terlihat jelas: penutur sedang membandingkan pilihan yang diambil (jalan kaki) dengan pilihan yang tidak diambil (naik taksi), dan menjelaskan secara spesifik apa yang hilang karena tidak memilih taksi

    Jika versi dengan `been`: menilai taksi itu sendiri apakah ide itu baik atau tidak, secara keseluruhan

    Tidak pakai `had`

    `They will have been surprised by the news` = `Mereka pasti sudah terkejut oleh berita itu`

    `He will have already left by now` = `Dia pasti sudah berangkat sekarang`
    - `Pasti` ini bertugas sebagai penanda keyakinan tinggi, tidak secara umum artinya ini
- Jika menggunakan `Will`

    Maknanya berubah total dari "penyesalan atas masa lalu" menjadi prediksi/keyakinan tentang sesuatu yang akan terbukti benar di masa depan, dilihat dari sudut pandang sekarang

    `That will have been the taxi driver who called earlier` = `Itu pasti sopir taksi yang menelepon tadi`

    `will have been` di sini menyatakan dugaan berdasarkan perhitungan/kepastian yang wajar.Penutur menyimpulkan sesuatu berdasarkan fakta yang sudah diketahui/diperkirakan sebelumnya <br>
    (misalnya, "aku memang sedang menunggu telepon dari sopir taksi, jadi itu pasti dia")

    Kesan: "Sudah pasti begitu, sesuai dugaanku"

    Menggunakan `must` $\to$ lebih umum digunakan

    `must have been` menyatakan dugaan berdasarkan bukti/deduksi logis. Penutur menyimpulkan sesuatu dari petunjuk yang teramati <br>
    (misalnya, nada suara di telepon, nomor yang tidak dikenal, waktu telepon masuk yang cocok)

    Kesan: "Berdasarkan bukti, satu-satunya penjelasan masuk akal"

    `Someone called and mentioned picking me up outside. That must have been the taxi driver who called earlier` <br>
    = `Ada seseorang yang menelepon dan bilang akan menjemput saya di luar. Pasti itu sopir taksi yang menelepon tadi` 
- Bentuk lainnya 

    Berbeda dengan yang di atas, ini kita belum memilih dan kita memberikan suatu saran

    `It's raining and traffic looks bad. Taking a taxi might be a good idea` <br>
    = `Sedang hujan dan lalu lintas terlihat buruk. Naik taksi mungkin ide yang bagus` (saran untuk sekarang)

##### **`I have come to...`**

Secara harfiah `have come` artinya `telah datang`, tapi ketika diikuti `to` maka maknanya seperti berproses hingga akhirnya...

`I have come to realize the truth` = 
- Makna literal: `Aku sekarang sampai pada titik di mana aku menyadari kebenaran`
- Makna alami: `Aku akhirnya menyadari kebenarannya`

Dengan menggunakan `i have come to` maka nuasannya seperti dulu saya belum menyadari lalu setelah melalui beberapa proses saya akhirnya sekarang menyadarinya

`I have come to enjoy cooking`
- Jika kita memahami `i hace come to` itu sebagai `akhirnya`: `Aku akhirnya menikmati memasak`
- Makna lebih alami
    - `Saya jadi menikmati kegiatan memasak`
    - `Lama-kelamaan aku jadi suka memasak`

Perbedaan dengan `I finally realized/realize the truth`

Tidak ada penekanan pada proses panjangnya

##### **Penyimpulan**

- `He must have been sleeping` = `Dia pasti sedang tidur`

    Fokusnya bukan hasil, tetapi kegiatan yang sedang berlangsung, yakni aktivitas sedang tidur

    (tidak melihat langsung, tetapi berdasarkan bukti atau situasi, dapat disimpulkan demikian)

    `must` dalam konteks ini artinya bukan harus, melainkan dugaan dengan keyakinan tinggi
- `He must have slept` = `Dia pasti sudah tidur`

    (Fokusnya pada fakta bahwa dia tidur telah terjadi dengan tidak menekankan proses)

___
Menggunakan `had`

had menunjukkan bahwa aktivitas tersebut sudah berlangsung sebelum kejadian lain di masa lalu

`He had been sleeping before I arrived` = `Dia sudah tidur sebelum saya tiba`

Saat aku datang, proses tidurnya sudah berlangsung (dalam artian lain sudah sudah tidur dan tidurnya sedang berlangsung)

##### **`have/had + Had`**

Kejadian terjadi di masa lalu, tetapi hasil atau dampaknya masih relevan sekarang.

Ini termasuk **Present Perfect**
- `Have + had`

    `I have had an accident` = `Aku telah mengalami kecelakaan`

    Dalam percakapaan, maknanya seperti "aku baru saja mengalami kecelakaan" (dampaknya masih ada terasa)

    Saya sudah/pernah mengalami atau memiliki sesuatu, dan hal itu relevan dengan situasi sekarang
    - Simple past

        Kemarin ada rapat yang aku tidak datang, esok harinya saya bertemu temen saya di kantor

        A: `Why were you absent yesterday?` = `Kenapa kamu absen kemarin?`

        B: `I had an accident yesterday` = `Saya mengalami kecelakaan kemarin`

        Menjelaskan alasan ketidakhadiranku kemarin, peristiwanya juga sudah selesai kemarin

        `We had great time in park` = `Kami bersenang-senang di taman`
    - Present Perfect

        Teman tiba-tiba menelpon karena hari ini ada rapat kerja di kantor

        A: `Where are you?`

        B: `I have had an accident, I can't come to work today` = `Saya mengalami kecelakaan, aku tidak bisa kerja hari ini`

        Aku mengalami kecelakaan, jadi sekarang aku tidak bisa datang. Ada akibat yang masih berlangsung, misal masih di rumah sakit, cedera, atau mobil rusak

        `He has had three meetings today` = `Dia telah menjalani tiga pertemuan hari ini`

        Past Perfect sangat sering digunakan untuk menjelaskan: "Sebelum kejadian yang sedang saya ceritakan, kondisi ini sudah terjadi."

    `I've had enough` = `Saya sudah tidak tahan` / `Saya sudah cukup`

    Nuansanya: "Sampai sekarang, saya sudah mencapai batas."

    `I've had a really bad day` = `Hari ini benar-benar buruk bagiku`

    Nuansanya sekarang: "Lihat kondisiku sekarang. Hari yang baru saja aku jalani itu buruk."
- `Had + had`

    Pada suatu titik dalam cerita masa lalu, saya sudah mengalami atau memiliki sesuatu

    `I'd had some strange experiences, so I wasn't surprised` <br>
    = `Saya pernah mengalami beberapa kejadian aneh, jadi saya tidak terkejut`

    Di sini `had had` berfungsi sebagai background/context untuk kejadian masa lalu

Secara sederhana, `... have had ...` dapat diartikan menjelaskan pengalaman atau kejadian yang pernah dialami (artinya sudah dialami) dan masih berkaitan dengan kondisi sekarang, sedangkan `... had had ...` menjelaskan pengalaman atau kejadian yang pernah dialami (artinya sudah dialami) sebelum atau pada suatu kondisi/kejadian di masa lalu.
- `have had`

    A: `Why don't you trust him?` = `Mengapa kamu tidak memercayainya?`

    B: `I've had some problems with him` = `Saya sempat mengalami beberapa masalah dengannya`

    Di suatu perusahaan, kamu masih bekerja di sana atau sedang memberikan pendapat tentang pengalamanmu sampai sekarang

    `I've had a great experience working here` = `Saya memiliki pengalaman yang sangat baik bekerja di sini`
- `had had`

    `I'd had some problems with him, so I decided to leave`

    Saya waktu itu sudah mengalami beberapa masalah dengannya, jadi saya memutuskan untuk pergi.

    Ini dalam konteks menceritkan sesuatu atau memberitahu kejadian yang sudah terjadi

    Setelah keluar kerja lalu mendapatkan pekerjaan baru, kemudian kita bercerita tentang perusaahn lama kita

    `I'd had a great experience working there before I moved` <br>
    = `Saya memiliki pengalaman yang sangat baik saat bekerja di sana sebelum saya pindah.`

#### **Causative**

- Active Causative

    $$ \boxed{\text{be + (have/get/let/make) + someone + v1}} $$
    Fokusnya pada siapa yang saya suruh atau tugaskan.
    - `have`

        Menugaskan atau mengatur

        `I have John clean the room` = `Saya menyuruh John membersihkan ruangan itu`

        Maknanya lebih mengarah bahwa pekerjaan itu memang dilaksanakan (Kesannya tindakan itu memang terjadi)

        Sedangkan `I told John to clean the room` (Artian memang mirip atau sama, namun makna dan nuansa berbeda) hanya mengarah ke kita menginstruksi (bisa jadi tidak jhon lakukan)
    - `Get`

        Berhasil membuat seseorang mau melakukan sesuatu dengan bujukan. Artinya sebelumnya belum tentu orang itu mau

        `I got John to clean the room`

        Ini menggunakan v2 artinya menceritakan (kejadiannya sudah lewat)

        `Get` sendiri memiliki banyak makna tergantung polanya
        - Mendapatkan

            `I get good grades` = `Saya mendapat nilai bagus`
        - Menjadi
            - v2

                `I got tired after work` = `Saya menjadi lelah setelah bekerja`

                Setelah bekerja teman anda bertanya `How was work yesterday?` atau bisa dalam konteks menceritakan pengalaman
            - v1

                `I get tired after work`

                Suasananya sedang menjelaskan kondisi umum. Misalnya saya memang tipe orang yang cepat lelah setelah kerja
        - Tiba

            `I get home at six` = `Saya sampai rumah jam enam`
    - `Let`

        Let bukan membuat seseorang melakukan sesuatu, tetapi mengizinkan seseorang melakukan sesuatu.

        `My parents let me stay out late` = `Orang tua saya mengizinkan saya pulang larut malam`


    - `Make`

        Dalam kondisi ini orangnya ada nuansa paksaan untuk melakukan sesuatu

        `My dad made me clean my room` = `Ayah menyuruhku membersihkan kamarku`

        ini versi dengan v2 karena sedang menceritakn (hal tersebut sudah terjadi)
- Passive Causative

    $$ \boxed{\text{be + have/get/let/make + something + v3}} $$

    Fokus pada hasil pekerjaan pada objek, tanpa harus menyebutkan siapa yang mengerjakannya

    Karena objek yang menerima tindakan menggunakan bentuk V3 (past participle)

    `I had my hair cut` = `Saya memotong rambut saya`

    Pelakunya (si pemotong) bahkan dihilangkan karena tidak penting.

    Yang penting adalah saya membuat rambut saya dipotong.

**Perbedaan**
- Biasa

    Active: `I repaired my car` = `Saya memperbaiki mobil saya`

    Passive: `My car was repaired` = `Mobil saya diperbaiki`
- Causative

    Active: `I had John repair my car` = `Saya meminta John memperbaiki mobil saya`

    Passive: `I had my car repaired` = `Saya membuat mobil saya diperbaiki`

#### **Other**

##### **V1 vs V2**

- Pemahaman native speaker tanpa menjelaskan eksplisit waktu
    - v1
    - `answers`

        Anda pegawai baru yang ingin tahu siapa yang bertanggung jawab mengangkat telepon.
        
        Anda bertanya: "Who `answers` customer calls?"
    - v2
        - `answerd`

            Anda sedang bekerja.

            Tiba-tiba telepon dari pelanggan penting masuk.

            Saat itu Anda sedang rapat, jadi tidak bisa mengangkatnya.

            Beberapa menit kemudian, Anda keluar dari ruang rapat dan bertanya: Who answered the call?

            Makna yang ingin Anda sampaikan: "Tadi, ketika telepon itu masuk, siapa yang mengangkatnya?"

            Yang ingin diketahui adalah satu kejadian yang sudah selesa
        - `cleaned`

            Hari ini ada pesta yang membuat ruangan jadi kotor, namun sekarang sudah bersih.

            Bos bertanya: "Who `cleaned` the room?"

            Yang ingin diketahui adalah siapa yang tadi membersihkan?
- Makna
    - v1

        `where do you work when you are doctor?` = `Di mana Anda bekerja jika Anda seorang dokter?`        

        Di sini yang dibahas adalah keadaan umum, bukan pengalaman seseorang.

        Ada banyak cara memahaminya, salah satunya dengan memahami kebiasaan atau rutinitas tempat tinggal bagi seseorang yang saat ini masih berstatus mahasiswa aktif, namun sedang tidak berada di kampus (misalnya saat liburan semester atau akhir pekan)
    - v2

        `Where did you work when you were doctor?` = `Di mana Anda bekerja saat Anda menjadi dokter?`

        Menanyakan lokasi tempat tinggal di masa lalu, di mana periode perkuliahan tersebut sudah selesai sepenuhnya
- **`Do` vs `Did`**
    - `Do`

        Situasi: Dua orang baru kenal di sebuah acara kumpul-kumpul (movie night gathering), sedang berbincang untuk saling mengenal minat masing-masing. Tidak ada film spesifik yang sedang dibicarakan — ini murni soal preferensi/kebiasaan
        - A: "So, what kind of movies are you into?"
        - B: "I like a lot of genres, actually."
        - A: "Do you watch horror movies?"
        - B: "Yeah, I do. I actually love jump scares."
    - `Did`

        Situasi: Dua teman sebelumnya sudah sepakat untuk menonton film tertentu (misalnya sudah direkomendasikan atau dijadwalkan). Sekarang salah satu bertanya apakah rencana itu sudah dilaksanakan
        - A: "Hey, I sent you that link to Hereditary last week."
        - B: "Oh right, I saw it."
        - A: "Did you watch the movie? What did you think?"
        - B: "Yeah, I did. It was terrifying, honestly."
- **`Has/Have` vs `Had`**
    - Biasa
        - `Has/Have`

            Tuan Sakti sedang wawancara kerja sebagai data scientist. Pewawancara ingin tahu pengalaman yang masih relevan hingga sekarang, tanpa menyebut kapan persisnya.
            - Interviewer: "Have you worked with large datasets before?"
            - Tuan Sakti: "Yes, I have. I've handled datasets with millions of rows using Python and MySQL."
            - Interviewer: "Has your team ever used Optuna for hyperparameter tuning?"
            - Tuan Sakti: "Yes, we have. We've integrated it into our MobileNetV2 pipeline."
        - `Had`

            Tuan Sakti menceritakan pengalaman lampau — misalnya, saat pertama kali menjalankan training model, ternyata sudah ada masalah sebelum proses training itu dimulai.

            "When I started training the MobileNetV2 model, I realized the protobuf version had caused a conflict with TensorFlow. I had already fixed similar issues before, so I knew how to solve it quickly"
    - Dengan been
        - `Has/Have`
            - 1

                Situasi: Rekan kerja Tuan Sakti menanyakan progres proyek penelitian polinomial ganjil yang sedang dikerjakan — aktivitas ini masih berlangsung sampai sekarang.
                - Rekan: "How's the research going?"
                - Tuan Sakti: "I've been working on the odd polynomial approximation project for about two months now. I've been comparing it with Newton-Schulz iteration."
                - Rekan: "Has your team been testing the convergence analysis too?"
                - Tuan Sakti: "Yes, we have. We've been running simulations every week."
            - 2

                Anda datang ke kantor. John masuk ke ruang rapat satu jam yang lalu yang sampai sekarang dia masih di dalam.

                Anda bertanya

                `Has John been cleaning the room?` = `Apakah John sudah membersihkan ruangan itu?`

                Yang ingin diketahui bukan spakah ruangannya sudah bersih?
                
                Tetapi apakah sejak tadi John sedang mengerjakan pekerjaan itu?
            - 3

                A: `The room looks perfect when everyone arrives`

                B: `Has John been cleaning the room?`

                Ini tidak menggunakan v2, jadi tidak secara spesifik membahas keadaan sebelumnya yang sudah terjadi atau bukan sedang menceritakan pesta kemarin.

                Tetapi sedang menjelaskan bagaimana acara biasanya berlangsung
            - 4

                `Has John actually been working on that task from a while ago until now?` <br>
                = `Apakah sejak tadi sampai sekarang John memang sedang mengerjakan pekerjaan itu?`
        - `Had`
            - 1

                Tuan Sakti bercerita tentang kondisi sebelum akhirnya berhasil menyelesaikan implementasi LoRA di PyTorch — ada proses panjang sebelum momen keberhasilan itu terjadi.

                "By the time I finally got the LoRA implementation working, I had been debugging the matrix decomposition for almost five hours. I had been trying different rank configurations before I realized the issue was in the weight initialization."
            - 2

                `Had John been cleaning the room before the guests arrived?` <br>
                = `Apakah John sedang membersihkan ruangan sebelum para tamu tiba?`

                Ini menayanakan kejadian dimasa lalu
                
                Sebelum tamu datang (dimasa lalu), apakah John memang sedang membersihkan selama beberapa waktu?

                Fokusnya pada proses
            - 3

                Bayangkan hari ini hari Senin.

                Tetapi mereka sedang membahas pesta hari Sabtu.

                A: `The room looked perfect when everyone arrived`

                B: `Had John been cleaning the room?`

                Sebelum momen yang sedang kita bicarakan, apakah John memang sedang sibuk membersihkan?

##### **`Either` | `Neither`**

Ini berlaku untuk 2 pilihan
- `Either`

    Maknanya menunjukkan bahwa minimal satu pilihan bisa dipilih, biasanya satu saja

    `You can either stay home or come with us` = `Kamu bisa tetap di rumah atau ikut dengan kami`

    `Either of these books is interesting` = `Salah satu dari buku-buku ini menarik`
- `Neither`

    Ini kebalikan dari either, artinya tidak satupun dari keduanya

    `Neither answer is correct` = `Tidak satu pun jawaban benar`

    `Neither of my friends came` = `Tidak satu pun temanku datang`

Jika lebih dari 2 maka gunakan `Any` (Salah satu dari banyak) dengan lawannya `None` (Tidak ada satupun dari banyal)

`Any book is interesting` = `Buku mana pun menarik` (Ada lebih dari 2 buku)

`None of the students passed` = `Tidak satu pun siswa lulus` (Ada lebih dari 2 murid)

##### **Modal Verbs**

- `Can` | `Could`
    - Kemampuan
        - `can`

            `I can swim` 

            Ini adalah kemampuan yang saya miliki sekarang
        - `could`

            `I could swim when I was five` = `Saya sudah bisa berenang saat berusia lima tahun`
    - Kemungkinan
        - `can`

            `AI can replace many jobs` $\to$ AI memang memiliki kemampuan menggantikan banyak pekerjaan (Nuansa lebih tegas)

        - `could`

            `AI could replace many jobs` $\to$ Ada kemungkinan AI menggantikan banyak pekerjaan
    - Perbedaan

        `One mistake can ruin everything`

        Satu kesalahan memang mampu merusak semuanya. Lebih berupa fakta umum.
        
        `One mistake could ruin everything`

        Ada kemungkinan satu kesalahan merusak semuanya
        
Kekuatan kemungkinan akan terjadi (paling besar ke paling kecil)
$$ \boxed{\text{Definitely > Will > May > Might > Impossible}} $$
- `may`

    - Meminta atau memberi izin

        Lebih formal daripada can
    
        A: `May I come in?` = `Bolehkah saya masuk?` <br>
        B: `Yes, you may` 

        A: `May I ask you a question?` = `Bolehkah saya bertanya?` <br>
        B: `No, you may not`
    - Menyatakan kemungkinan
        - Biasa

            `It may rain tomorrow` = `Mungkin besok akan hujan`

            `She may be at home` = `Dia mungkin ada di rumah`
        - Kemungkinan sedang terjadi

            `She may be sleeping` = `Mungkin dia sedang tidur`
        - Kemungkinan terjadi dimasa lalu

            `He may have forgotten` = `Dia mungkin sudah lupa`
- `might`

    Secara tata bahasa hampir sama dengan may, tetapi maknanya lebih lemah.
- `should`

    Makna utamanya: sebaiknya, semestinya, atau seharusnya

    `You should study harder` = `Kamu sebaiknya belajar lebih giat`

    Kalimat ini berupa saran, jika menggunakan `must` maka ada kesan perintah disitu

    `The train should arrive soon` = `Kereta seharusnya segera datang`

    - `+ v3`

        Digunakan untuk menyatakan penyesalan, kritik, atau kewajiban yang tidak dilakukan di masa lalu.
    
        `You should have called me` = `Seharusnya kamu meneleponku`
- `ought to`

    Berbeda dengan modal lain `(can, may, should, must)` yang langsung diikuti V1, ought selalu diikuti to + V1

    Penggunana memang mirip `should`, namun `ought to` sendiri lebih menekankan kewajiban moral, hal yang benar, apa yang pantas dilakukan, atau ekspektasi yang masuk akal

    `You ought to apologize` = `Anda sebaiknya meminta maaf`

    `You ought to respect your parents` = `Kamu sebaiknya menghormati orang tuamu`

##### **Past Modals**

Bentuk modal verbs (can, could, may, might, must, should, will, would, dll.) yang digunakan untuk membicarakan kemungkinan, kewajiban, atau spekulasi terkait masa lalu
- Spekulasi/Deduksi tentang Masa Lalu

| **Modal**                        | **Fungsi**                       | **Contoh**                                                                  |
| ---------------------------- | ---------------------------- | ----------------------------------------------------------------------- |
| **must have + V3**           | Kepastian tinggi (positif)   | *She must have left early.* (Saya yakin dia sudah pergi lebih dulu.)    |
| **can't/couldn't have + V3** | Kepastian tinggi (negatif)   | *He can't have finished already.* (Mustahil dia sudah selesai.)         |
| **may/might have + V3**      | Kemungkinan (tidak pasti)    | *They might have missed the bus.* (Mungkin mereka ketinggalan bus.)     |
| **could have + V3**          | Kemungkinan yang lebih lemah | *It could have been an accident.* (Mungkin saja itu sebuah kecelakaan.) |

- Penyesalan atau Kritik terhadap Masa Lalu

### **Relative Pronouns**

Merupakan kata ganti penghubung ketika digunakan untuk menghubungkan dua klausa. Namun, kata-kata ini juga dapat berfungsi sebagai interrogative pronouns (kata tanya)

- Who

    Digunakan ketika kata yang digantikan adalah orang dan berperan sebagai subjek

    `The boy who won the competition is my brother`

    Yang memenangkan lomba adalah `the boy`, berarti dia pelaku.

    Jika yang melakukan aksi adalah orang maka gunakan `who`
- Whom (Formal)

    `The teacher called the student` = Siswa yang dipanggil oleh guru
    - `teacher`: Pelaku
    - `student`: Objek

    `The student whom the teacher called`

    Menggunakan `whom` karena `teacher` yang melakukan aksi, jika versi di atas pertama itu yang meme

    `The boy whom I congratulated won the competition` = Anak laki-laki yang saya beri ucapan selamat itu memenangkan kompetisi

    Penggunana `whom` itu kadang bisa digantikan oleh `who`, tapi ada beberapa kasus penggunana `who` tidak bisa digantikan oleh `whom` jika `who` berfungsi sebagai subjek
    - `The student whom the teacher called`: Bisa digantikan dengan `who`
    - ``The boy who won the competition is my brother``: Tidak bisa digantikan dengan `whom`, Karena `who` pada kalimat tersebut berfungsi sebagai subjek
- Which

    Dipakai untuk benda

    `The book which I bought is interesting` = `Buku yang aku beli itu menarik`

    `The movie which won the Oscar` = `Film yang memenangkan oscar`

    Menggunakan v2 karena pemenang oscar telah ditunjuk, ini dalam konteks membahas siapa yang menang. 
    
    Jika mau menggunakan v1 `(wins)` maka konteksnya bukan membicarakan satu kemenangan yang sudah selesai, melainkan sebuah fakta umum.

    `The movie which wins the Oscar usually becomes famous` = `Film yang memenangkan Oscar biasanya menjadi terkenal`

    Fokusnya pada fakta umum, bukan fokus pada suatu film yang memenangkan oscar (kejadian sudah selesai lalu ini dibicarakan)
- Whose

    Menunjukkan kepemilikan

    `I know the girl whose father is a doctor` = `Aku tau perempuan yang ayahnya seorang dokter`

    `Whose car is this?` = `Mobil siapa ini?`

### **Penghubung**

#### **`Conjunction`: Penghubung Antarklausa**

##### **`As`**

- Sebagai (funsgi atau peran)

    `She works as a teacher`
- Saat/ketika

    Menunjukkan 2 kejadian terjadi secara bersamaan

    `As i was walking it, it started to rain` = `Saat aku sedang berjalan, hujan mulai turuna`

    Pembicara sedang menceritakan kembali kejadian yang sudah selesai atau seperti menceritakan pengalaman.

    Versi menggunakan `am` dan `v1`

    `As I am walking, it starts to rain` = Saat aku sedang berjalan (sekarang), hujan mulai turun

    Pembicara seolah sedang berada di dalam kejadian tersebut.
- Karena
    - Jika dalam keadaan sedang nongkrong sama teman dan mau pulang maka

        `As it is late, I'm going home` = `Karena sudah larut, saya akan pulang`
    - Dalam keadaan kemarin saya pulang duluan dari nongkrong lalu dihari kemudian temen saya bertanya kenapa pulang
        - Fokus pada kejadian yang sudah selesai (hasil)
        
            `As it was late, i went home` = `Karena sudah larut malam, aku mau pulang`
        - Fokus pada proses yang sedang terjadi

            `As it was late, I was going home when John called me` = `Karena sudah larut, saya sedang dalam perjalanan pulang ketika John menelepon`

            Di sini `was going` dipakai karena ada kejadian lain (John called me) yang memotong aksi yang sedang berlangsung


    Jika seseorang bertanya "Kenapa kamu pulang?", kita biasanya ingin menjelaskan alasan secara langsung, sehingga because adalah pilihan yang paling alami.

    Jika kita sendiri yang memutuskan untuk pulang dan ingin memberi tahu orang lain, kita biasanya hanya memberikan latar belakang atau konteks keputusan tersebut, sehingga as terdengar lebih alami (meskipun because juga tetap benar).

| As                                                                      | Because                                                     |
| ----------------------------------------------------------------------- | ----------------------------------------------------------- |
| Menyatakan alasan yang dianggap sudah jelas atau sebagai latar belakang | Menyatakan alasan utama atau penyebab yang ingin ditekankan |
| Fokus pada keadaan/situasi                                              | Fokus pada sebab-akibat                                     |
- Seperti

    `Do it as i showed you` = `Lakukan seperti yang aku tunjukkan padamu` 

    menggunakan v2 karena sebelumnya sudah ditunjukkan sesuatu lalu sekrang kita meminta seseorang melakukannya
- Meskipun (formal)

    Di sini as digunakan sebagai concessive as, fungsinya untuk menunjukkan kontras

    `Strong as he is, he could not lift the box` = `Mesikupn dia kuat, di tidak bisa menggangkat kotak tersebut`

    Menggunakan `could` karena sedang menceritakan kejadian yang sudah terjadi (dia sudah mencoba menggangkat kota)
- Sebagai mana/seperti yang

    `as you know, the meeting is tomorrow` = `seperti yang kamu tahu, rapatnya besok`

    `as i said earlier, this is important` = `Seperti yang saya (sudah) katakan, ini penting`
- Sebanding dengan

    perbandingan kesetaraan (equality comparison)

    `He is as smart as his father` = `Dia sama cerdasnya dengan ayahnya` atau `Dia sepintar ayahnya`

    `This is as good as it gets` = 
    - `Ini adalah yang terbaik dari yang terbaik`
    - `Ini sudah sebaik mungkin`
    - `Ini adalah kondisi terbaik yang bisa didapat`
- Seakan-akan
    - Bertentangan dengan kenyataan

        $$ \text{S + v + as if + S + v2} $$
        - $\text{v2}$ : Ini bukan menunjukkan waktu lampau, melainkan mengisyaratkan situasi yang dibayangkan tidak sesuai dengan kenyataan.

        `He talks as if he knew everything` = `Dia berbicara seolah-olah dia tahu segalanya (padahal tidak)`
    - Dugaan berdasarkan apa yang diketahui

        `it looked as though it would rain` =
        - `Sepertinya akan turun hujan`
        - `Dari apa yang terlihat saat itu, tampaknya hujan akan turun`

        Saya sudah melihat sebelumnya lalu mencoba menjelaskan 

        Penggunana `as though` sama seperti `as if` di atas, hanya saja dengan `though` lebih formal 

        Keduanya juga bisa diartikan `seperti`
- Sesuai dengan (formal)

    `As per our agreement, payment is due today` = `Sesuai dengan kesepakatan kita, pembayaran jatuh tempo hari ini`
    - `As per` = `Sesuai dengan`

    Makna yang sama, namun nuansa lebih santai

    `According to our agreement, payment is due today`
- Mengenai/tentang

    `As to the report, it is not ready yet` = `Mengenai laporan itu, laporan tersebut belum siap`    

##### **`while`**

- sambil / sementara

    Konsepnya dua aktivitas berjalan bersamaan

    `I studied while listening to music` = `Saya belajar sambil mendengarkan musik`

    `She cooked while I washed the dishes` = `Dia memasak sementara saya mencuci piring`
- Sedangkan

    Membandingkan dua hal

    `John likes coffee, while Mary prefers tea` = `John menyukai kopi, sedangkan Mary lebih memilih teh`

`whereas` itu hampir sama dengan `while`, tetapi lebih formal dan lebih menekankan kontras.

#### **`Conjunctive Adverb`: Penghubung Ide atau Kalimat**

##### **`However`**

However bukan conjunction (kata penghubung antarklausa), melainkan conjunctive adverb atau discourse marker, yaitu kata yang menghubungkan ide atau kalimat, bukan menghubungkan dua klausa secara langsung seperti but, although, atau while.

Konsep dasarnya adalah <br>
Informasi kedua bertentangan atau berlawanan dengan apa yang diharapkan dari informasi pertama

`John studied hard. However, he failed` = `John belajar dengan giat. Namun, dia gagal`

But menghubungkan dua klausa dalam satu kalimat.

`I wanted to go, but I was busy` = `Saya ingin pergi, tetapi saya sibuk`

However menghubungkan dua kalimat atau dua ide.

`I wanted to go. I was, however, busy` = `Saya ingin pergi. Namun, saya sibuk`

Ini salah $\to$ `I wanted to go, however I was busy` lalu yang benar `I wanted to go. However, I was busy` Karena however bukan conjunction

`He is rich. However, he isn't happy` = `Dia kaya. Namun, dia tidak bahagia`

- `I got you`: Aku paham / Aku mengerti / Aku akan membantumu / Tenang, aku tangani

    **I get you** dan **I got you** sama-sama dapat berarti “aku paham kamu/maksudmu”, tetapi nuansa dan situasinya berbeda. **I get you** menggunakan *get* dalam arti “memahami” dan biasanya menekankan bahwa pembicara **memahami cara berpikir, perasaan, atau maksud seseorang**. Karena itu, ungkapan ini sering muncul ketika seseorang sedang menjelaskan perasaan atau sudut pandangnya: *“I get you. You don't want to hurt her feelings.”* = “Aku paham kamu. Kamu tidak ingin menyakiti perasaannya.” Nuansanya lebih seperti **“Aku mengerti apa yang kamu maksud/rasakan”**, sehingga sering memiliki unsur empati atau menunjukkan bahwa kita benar-benar memahami orang tersebut. Sementara itu, **I got you** dapat berarti “aku sudah paham/menangkap maksudmu”, terutama sebagai **respons langsung setelah seseorang memberikan penjelasan atau instruksi**: *“First, click this button and then select the file.” — “I got you.”* = “Oke, aku paham.” Jadi, *I got you* terasa lebih seperti **“oke, sudah kutangkap”** atau “siap”, bukan sekadar pernyataan bahwa kita memahami seseorang.

    Perbedaan lainnya adalah **I got you memiliki makna yang jauh lebih luas daripada I get you**. Dalam situasi tertentu, *I got you* bukan berarti “aku paham”, melainkan **“aku akan membantumu”, “aku tangani”, “aku menjagamu”, atau “kamu bisa mengandalkanku”**. Misalnya: *“Can you help me with this?” — “I got you.”* = “Bisa, aku bantu.” Atau: *“Don't worry, I got you.”* = “Tenang, aku akan membantumu/menjagamu.” Bahkan dalam konteks pembayaran, *“Don't worry, I got you”* dapat berarti “Tenang, aku yang bayar.” Jadi secara praktis, **I get you** terutama berfokus pada **understanding seseorang**, sedangkan **I got you** lebih kontekstual dan dapat menunjukkan **pemahaman, kesiapan bertindak, bantuan, dukungan, atau perlindungan**. Cara mudah mengingatnya: **I get you = “aku memahami kamu”**, sedangkan **I got you = “oke, aku sudah menangkapnya / tenang, aku urus atau bantu kamu.”**

#### **`be` | `being` | `been`**

- `be`

    Makna dasarnya bukan hanya "adalah", tetapi lebih tepat dipahami sebagai menjadi, berada, atau dalam suatu keadaan

    `I want to be a doctor` = `Saya ingin menjadi dokter`

    `I hope to be successful` =
    - `Saya berharap bisa sukses`
    - `Saya berharap menjadi sukses`

    `This needs to be fixed` = `Hal ini perlu diperbaiki`
- `being`

    Makna dasarnya adalah sedang berada, sedang menjadi, atau proses berada dalam suatu keadaan
    - Present/past Continuous Passive

        $$ \boxed{\text{be + being + V3}} $$
        - 1

            `The house i being painted right now` = `Rumah itu sedang dicat sekarang` (tidak membahas siapa yang mengecat)

            Kalimat ini menggambarkan situasi seperti:

            Anda melewati sebuah rumah

            Anda melihat: tukang sedang mengecat, ada ember cat, ada tangga, ada kuas, dan sebagian tembok belum selesai

            prosesnya itu sedang Anda amati.
        - 2

            `The car was being repaired when i arrived` = `Mobil itu sedang diperbaiki ketika saya tiba`

            ini kan menggunakan `was` dan `arrived` yang merupakan v2

            Menggunakan v2 sebagai menceritakan atau membahas waktu sebelumnya, jika pakai v1 itu membahas fakta umum atau kebiasana

            Saya menceritakan kejadian yang **telah terjadi**. Saat saya tiba, proses perbaikan sedang berlangsung.

            Mengapa memakai "arrived" (V2)? <br>
            Karena arrived adalah kejadian yang terjadi sekali (completed event).
    - Gerund

        `being` dipakai sebagai subjek

        `Being honest is important` = `Bersikap jujur itu penting`

        `I appreciate you being honest with me` = `Saya menghargai kejujuran Anda kepada saya`
    - Menunjukkan Perilaku

        `He is being silly` = `Dia sedang bertingkah konyol`
- `been`

    Makna dasarnya sudah berada, sudah menjadi, pernah berada, atau telah berada dalam suatu keadaan

    Karena bentuk V3, been selalu dipakai bersama auxiliary seperti have, has, had, will have.

#### **`Bear`**

Kata bear memang paling terkenal sebagai kata benda yang berarti beruang

Penggunaan Bear sebagai verb:
- Menanggung / Memikul
    - I can't bear this pain = Aku tidak tahan rasa sakit ini

        Kalimat ini menyatakan keputusasaan yang amat sangat akibat beban berat, baik berupa rasa sakit fisik yang teramat parah maupun luka batin yang mendalam
    - She bears a heavy responsibility = Dia memikul tanggung jawab besar
- Tahan / Sanggup Menghadapi
    - I can't bear waiting any longer = Aku tidak tahan menunggu lebih lama
    - He couldn't bear to see her cry = Dia tidak sanggup melihatnya menangis

        Karena menggunakan could ini fokus ke pengalaman yang sudah berlalu, jadi ini sedang menceritakan
- Menunjukkan / Memiliki
    - The document bears his signature = Dokumen itu memuat tanda tangannya
- Menghasilkan
    - This investment may bear results = Investasi ini mungkin menghasilkan hasil
- Mengingat / Mempertimbangkan

    Bear in mind that = ingatlah / pertimbangkan bahwa

Perbedaan Bear dan Have
- Bear mengandung ide bahwa sesuatu itu melekat, terlihat, dibawa ke mana pun, bahkan menjadi bagian dari identitas atau keadaan subjek
    - The wall bears cracks 
        - Dinding itu menanggung retakan 
        - Dinding tersebut menunjukkan/memiliki retakan

        Retakan menjadi tanda yang tampak pada dinding

        Secara makna kiasan (figurative meaning) dalam sastra atau puisi, frasa ini sering dipakai untuk melambangkan:
        - Kerentanan: Struktur yang tampak kuat ternyata mulai rapuh dari dalam.
        - Sejarah atau Beban: Dinding diibaratkan menanggung beban berat, tekanan waktu, atau usia tua sehingga meninggalkan bekas luka berupa retakan.
        - Petunjuk Kerusakan: Ada masalah tersembunyi, trauma, atau tekanan yang sudah merusak keutuhan sesuatu.
- Have/have hanya mengatakan subjek mempunyai sesuatu.
    - The wall has cracks = Dinding itu memiliki retakan
    
        Hanya memberikan informasi kemepilikan retakan pada dinding

#### **`Increase` | `Enhance`**

keduanya sering diterjemahkan sebagai meningkatkan, tetapi maknanya berbeda. Perbedaannya terletak pada apa yang ditingkatkan.
- Increase ↔ Decrease
- Enhance ↔ Reduce

| Kata         | Makna utama                                                                              | Fokus                    |
| ------------ | ---------------------------------------------------------------------------------------- | ------------------------ |
| **increase** | Menambah jumlah, ukuran, nilai, tingkat, atau kuantitas                                  | Kuantitas (lebih banyak) |
| **enhance**  | Membuat sesuatu menjadi lebih baik, lebih efektif, lebih menarik, atau lebih berkualitas | Kualitas (lebih baik)    |

- Increase = meningkatkan jumlah atau besaran

    Increase berarti ada pertambahan secara kuantitatif
    - Sales increased by 20%: Penjualan meningkat 20%
    - The population is increasing: Populasi sedang bertambah
    - Increase the learning rate: Tingkatkan learning rate

    Di sini yang berubah adalah besar nilainya
- Enhance = meningkatkan kualitas

    Enhance berarti membuat sesuatu lebih baik, meskipun jumlahnya tidak berubah.
    - AI enhances decision making: AI meningkatkan kualitas pengambilan keputusan
    - This algorithm enhances model performance: Algoritma ini meningkatkan performa model

    Di sini bukan berarti jumlahnya bertambah, tetapi mutunya meningkat

#### **`Invent` | `Create`**

Keduanya sama-sama dapat diartikan sebagai "menciptakan" atau "membuat"
- `Invent`

    `Alexander Graham Bell invented the telephone` = `Alexander Graham Bell menemukan telepon`

    `Someone invented the smartphone` = `Seseorang menciptakan ponsel pintar`

    Fokusnya adalah siapa yang pertama kali menciptakan smartphone
- `Create`

    Bisa berarti: membuat, menghasilkan, menciptakan, membangun, atau menyebabkan

    B: `Yes she did, She created a beautiful painting` = `Ya dia yang melukisnya, Dia membuat lukisan yang indah`

    Lukisan itu baru, tetapi kegiatan melukis sendiri bukan penemuan

    **Jika menggunakan v1**

    `She creates beautiful paintings every year` = `Dia menciptakan lukisan-lukisan yang indah`

    Fokusnya: setiap tahun dia membuat satu lukisan yang indah. Tidak mengacu ke satu lukisan yang telah ia buat, kemudian jika ada seseorang yang bertanya lukisan tersbeut maka kita bisa jawab bahwa itu lukisan dia

            We go through this tunnel every day = Kami melewati terowongan ini setiap hari (Rutinitas)

            **Kalimat ini hanya mengatakan bahwa kamu mengalami masa sulit, tidak ada informasi apakah kamu berhasil mengatasinya**
        - Melewati tempat

            We went through the tunnel yesterday = Kemarin kami melewati terowongan itu (Selesai)
        - Memeriksa
        - Menjalani proses

    - `got`

        I got through a lot of stressful situations during college = Aku berhasil melewati banyak situasi yang membuat stres selama masa kuliah (Ini dalam konteks menceritakan pengalaman yang sudah selesai di masa lalu atau sudah selesai)

    I'll look through your report = Aku akan memeriksa laporanmu.

    Membaca dari halaman 1 sampai selesai

    Please read through the instructions = Silakan baca instruksinya sampai selesai
- `Think`

    Memikirkan secara matang, berpikir melewati seluruh kemungkinan hingga mendapatkan keputusan.

    Think through your decision = Pikirkan keputusanmu dengan matang

#### **`Used to`**

- `used to + V1`

    Kebiasaan atau keadaan di masa lalu yang sudah tidak berlaku sekarang

    `dulu pernah/biasa..., tetapi sekarang tidak lagi`

    `I used to play football` = `Dulu saya biasa bermain sepak bola`

    Fokusnya bukan waktu lampau biasa, tetapi perubahan antara masa lalu dan sekarang, dulu bias bermain sepak bola tapi sekarang tidak

    `There used to be a cinema here` = `Dulu ada bioskop di sini`

    Perbedaan dengan simple past itu hanya memberitahu bahwa kegiatan itu terjadi di masa lalu. Tidak memberi tahu apakah sekarang masih bermain.
- `used to + noun / V-ing`

    sudah terbiasa dengan sesuatu.

    sesuatu itu tidak terasa aneh, sulit, atau mengejutkan lagi
    - `Noun`
    
        `I am used to cold weather` = `Saya terbiasa dengan cuaca dingin`

        Bukan berarti saya suka dan bukan berarti cuaca dingin sudah selesai.

        Tetapi Cuaca dingin bukan masalah lagi bagi saya. Saya sudah terbiasa dengan cuaca dingin
    - `V-ing`

        `I'm used to waking up at 5 a.m` = `Saya sudah terbiasa bangun jam 5 pagi`
- `get used to + noun / V-ing`

    Fokusnya adalah proses adaptasi

    Misal, hari pertama jadi ceo semuanya terasa sulit. Seminggu kemudian mulai kurang sulitnya lalu sebulan kemudian mulai nyaman. Selama masa itu kita berkata:

    `I'm getting used to my new job` = `Saya mulai terbiasa dengan pekerjaan baru saya`

    Belum sepenuhnya nyaman. Masih proses.

    Setelah enam bulan.

    `I'm used to my new job` = `Saya sudah terbiasa dengan pekerjaan baru saya`

    Sekarang sudah terbiasa

#### **`Wish`**

- Sekarang

    $$ \boxed{\text{Subject + wish + Subject + V2 / were}} $$

    Digunakan ketika mengharapkan keadaan sekarang berbeda dari kenyataannya.

    Bentuknya memang memakai past tense, tetapi bukan berarti membicarakan masa lalu.

   **Past tense di sini berfungsi sebagai unreal past, yaitu menunjukkan sesuatu yang tidak nyata**.

    `I wish I were rich` = `Andai saja saya kaya`

    `I wish I had a car` = `Andai saja aku punya mobil`

    `I wish I knew the answer` = `Andai saja saya tahu jawabannya`
- Masa lalu

    $$ \boxed{\text{Subject + wish + Subject + had + V3}} $$

    Ini digunakan ketika menyesali sesuatu yang sudah terjadi di masa lalu.

    Peristiwa itu sudah selesai dan tidak bisa diubah.

    Mengapa memakai Past Perfect (had + v3)? <br>
    Karena kita sedang melihat masa lalu dari sudut pandang sekarang.

    `I wish I had studied harder` = `Andai saja dulu saya belajar lebih giat`

    `I wish I had gotten up earlier` = `Andai saja saya bangun lebih awal`
- Harapan untuk berubah 

    $$ \boxed{\text{Subject + wish + Subject + would + V1}} $$

    Di sini kita tidak sedang menyesali fakta, tetapi ingin ada perubahan perilaku atau kejadian

    Mengapa bukan Past? <br>
    Karena masalahnya bukan fakta, masalahnya adalah saya ingin sesuatu berubah mulai sekarang atau nanti.

    `I wish he would arrive on time` = `Saya berharap dia tiba tepat waktu`

    `I wish it would stop raining` = `Saya berharap hujannya berhenti`

    Perbedaan
    - Wish + past: `I wish the room were cooler` = `Andai saja ruangan ini lebih dingin`

        Fokus pada keadaan ruangan

        Dalam konteks ini `wish` tidak diartikan sebagai harapan semoga tetapi pengandaian sesuatu yang tidak nyata

        $\cancel{\text{i wish the room is cooler}} \to$ jika kita ganti `were` menjadi `is` atau dalam konteks lainnya yg sama ini menjadi salah, karena `the room is cooler` itu sendiri berupa fakta 
    - Wish + Would: `I wish the AC would work` = `Saya berharap AC ini mau berfungsi`

        Fokus pada perubahan

- `Certain`

    Menekankan bahwa ada suatu kelompok atau beberapa hal, tetapi identitasnya tidak penting atau tidak disebutkan
    - Determiner 

        certain + noun

        Makna umumnya tidak semua, sebagian yang memiliki karakteristik tertentu, atau kelompok tertentu

        - `particular`: menekankan satu hal atau satu anggota tertentu yang menjadi fokus.

            `This particular book is difficult` = `Buku yang satu ini sulit`

    - Adjective
